## Invalid Usage Values


```python
usage_columns = ["calls_made", "sms_sent", "data_used"]

for col in usage_columns:
    print(f"\n{col}")
    print("Negative:", (df[col] < 0).sum())
    print("Zero:", (df[col] == 0).sum())
    print("Positive:", (df[col] > 0).sum())
```

```python
for col in usage_columns:
    negative_pct = (df[col] < 0).mean() * 100
    print(f"{col}: {negative_pct:.2f}% negative")
```

```python
for col in usage_columns:
    print(f"\n{col}")
    print(
        df.groupby(df[col] < 0)["churn"]
        .agg(["count", "mean"])
    )
```

```python
invalid_usage = (
    (df["calls_made"] < 0) |
    (df["sms_sent"] < 0) |
    (df["data_used"] < 0)
)

print("Customers with at least one invalid usage value:", invalid_usage.sum())
print("Percentage:", invalid_usage.mean() * 100)
```
